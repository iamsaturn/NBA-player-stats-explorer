```python
notnull = df.loc[(~(df['Unnamed: 0'].isna())) & (~(df['Unnamed: 1'].isna())) & (~(df['Unnamed: 2'].isna()))]

if notnull.empty:
    df = df.drop(columns=['Unnamed: 0', 'Unnamed: 1', 'Unnamed: 2'])
```

```python
print(df.shape)
print(df.columns)
```

```python
keldon_games = df.loc[((df['Player'] == 'Keldon Johnson'))]
keldon_games['Date'].nunique()
```

```python
df_by_player = df.groupby('Player')[['PTS','AST','TRB']].mean().round(2)
df_by_player.head()
```

```python
games = df.groupby('Player').size()
print(type(games))
df_by_player['Games'] = games
df_by_player.head()
```

```python
fg_sum = df.groupby('Player')['FG'].sum()
fga_sum = df.groupby('Player')['FGA'].sum()
fg_pct = (fg_sum/fga_sum *100).round(2)
df_by_player['FG%'] = fg_pct
df_by_player.head()


print(df_by_player['FG%'].isna().sum())
```

```python
sum_3pa = df.groupby('Player')['3PA'].sum()
sum_3p = df.groupby('Player')['3P'].sum()
pct_3p = (sum_3p/sum_3pa *100).round(2)
df_by_player['3P%'] = pct_3p
df_by_player.head()

print(df_by_player['3P%'].isna().sum())
```

```python
sum_FTA = df.groupby('Player')['FTA'].sum()
sum_FT = df.groupby('Player')['FT'].sum()
pct_FT = (sum_FT/sum_FTA *100).round(2)
df_by_player['FT%'] = pct_FT
df_by_player.head()

print(df_by_player['FT%'].isna().sum())
```

```python
def minutes_to_decimal(tempo: str) -> float:
    partes = tempo.split(':')
    partes[0] = int(partes[0])
    partes[1] = int(partes[1]) / 60
    partes = sum(partes).round(2)
    return partes

df_minutes_to_decimal = df['MP'].apply(minutes_to_decimal)
df_minutes_to_decimal.head()
```
